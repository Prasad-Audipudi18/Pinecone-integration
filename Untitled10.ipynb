```python
from pinecone import Pinecone, ServerlessSpec
from sentence_transformers import SentenceTransformer
import time

api_key="pcsk_6pAxLj_UTdZAAyaFaXP5913FBsJMXLuqYUQcdTbPmiUZZxyhuC7yE7C62RhKaF6FiTxxxp"
texts=[
    "Ai is transforming health care",
    "Robotics is industrial automation",
    "Machine learning in finance",
    "Ai for simulation and modeling",
    "Cloud computing basics"
]
pc = Pinecone(api_key=api_key)
index_name = "quickstart"
if index_name not in pc.list_indexes().names():
    pc.create_index(
        name=index_name,
        dimension=384,
        metric="cosine",
        spec=ServerlessSpec(
            cloud="aws",
            region="us-east-1"
        )
    )
index = pc.Index(index_name)
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
for i,text in enumerate(texts):
   embedding=model.encode(text).tolist()
   metadata={
      "category": "ai" if "AI" in text else "other",
      "created_at":int(time.time()),
      "keywords": text.lower().split()
  }
  #  index.upsert([(f"doc={i}", embedding, metadata)])
  #  index.upsert([(f"doc={i}", embedding, metadata)])
   index.upsert(vectors=[(f"doc={i}", embedding, metadata)])
```

```python
query="Ai in simulations"
query_embedding= model.encode(query).tolist()
results=index.query(
    vector=query_embedding,
    top_k=3,
    include_metadata=True
)
print(results)
for match in results["matches"]:
   print("Score:", match["score"])
   print("Text:", match["metadata"]["keywords"])
```
